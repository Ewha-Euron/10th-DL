## 5장. 토큰화

### 예제 5.1 단어 토큰화 (Word Tokenization)
- `split()` 메서드를 사용하여 공백 기준으로 단어 분리
- 가장 간단한 방법이지만 '최고!'와 '최고'를 다른 토큰으로 처리

```python
# 예제 5.1 단어 토큰화
review = "현실과 구분 불가능한 CGI. 시각적 즐거움은 최고! 더불어 OST는 더더욱 최고!!"

tokenized = review.split()

print("단어 토큰화 결과:")
print(tokenized)
print(f"\n토큰 수: {len(tokenized)}개")
```

### 예제 5.2 글자 토큰화 (Character Tokenization)
- `list()` 함수를 사용하여 각 글자 단위로 분리
- 공백도 토큰으로 처리됨
- 한글은 자음+모음 조합이므로 자소 단위 토큰화도 가능

```python
# 예제 5.2 글자 토큰화
review = "현실과 구분 불가능한 CGI. 시각적 즐거움은 최고!"

tokenized = list(review)

print("글자 토큰화 결과:")
print(tokenized)
print(f"\n토큰 수: {len(tokenized)}개")
print("\n※ 단어 토큰화(13개)보다 훨씬 긴 시퀀스 생성됨")
```

### 예제 5.3 자소 단위 토큰화 (jamo 라이브러리)
- 한글을 초성·중성·종성 자소 단위로 분리
- `h2j()`: 완성형 한글 → 조합형 변환
- `j2hcj()`: 조합형 → 자소 단위 분리

```python
!pip install jamo
```

```python
# 예제 5.3 자소 단위 토큰화
from jamo import h2j, j2hcj

review = "현실과 구분 불가능한 CGI. 시각적 즐거움은 최고!!"

decomposed = j2hcj(h2j(review))
tokenized = list(decomposed)

print("자소 단위 토큰화 결과:")
print(tokenized)
print(f"\n토큰 수: {len(tokenized)}개")
print("\n※ 글자 토큰화보다도 더 긴 시퀀스 생성 (자음+모음으로 분리됨)")
```

```python
# 예제 5.4 Okt 형태소 분석 (Open Korean Text)
from konlpy.tag import Okt

okt = Okt()
sentence = "무엇이든 상상할 수 있는 사람은 무엇이든 만들어 낼 수 있다."

nouns   = okt.nouns(sentence)   # 명사 추출
phrases = okt.phrases(sentence) # 어절·구 추출
morphs  = okt.morphs(sentence)  # 형태소 추출
pos     = okt.pos(sentence)     # 품사 태깅

print("명사 추출 :", nouns)
print("구   추출 :", phrases)
print("형태소 추출:", morphs)
print("품사 태깅  :", pos)
```

```python
# 예제 5.5 Kkma 형태소 분석 (꼬꼬마 — 세종 말뭉치 기반)
from konlpy.tag import Kkma

kkma = Kkma()
sentence = "무엇이든 상상할 수 있는 사람은 무엇이든 만들어 낼 수 있다."

nouns     = kkma.nouns(sentence)     # 명사 추출
sentences = kkma.sentences(sentence) # 문장 분리
morphs    = kkma.morphs(sentence)    # 형태소 추출
pos       = kkma.pos(sentence)       # 품사 태깅

print("명사 추출 :", nouns)
print("문장 추출 :", sentences)
print("형태소 추출:", morphs)
print("품사 태깅  :", pos)
```

```python
# 예제 5.7 영문 토큰화 (단어 / 문장)
from nltk import tokenize

sentence = "Those who can imagine anything, can create the impossible."

word_tokens = tokenize.word_tokenize(sentence)  # 단어 토큰화
sent_tokens = tokenize.sent_tokenize(sentence)  # 문장 토큰화

print("단어 토큰화:", word_tokens)
print("문장 토큰화:", sent_tokens)
```

```python
# 예제 5.8 영문 품사 태깅 (Averaged Perceptron Tagger)
from nltk import tag, tokenize

sentence = "Those who can imagine anything, can create the impossible."
word_tokens = tokenize.word_tokenize(sentence)

pos = tag.pos_tag(word_tokens)

print("품사 태깅 결과:")
print(pos)
print("\n주요 품사 태그: NN=명사, VB=동사, JJ=형용사, RB=부사, PRP=인칭대명사")
```

```python
# 예제 5.12 SentencePiece BPE 토크나이저 모델 학습
import os
from sentencepiece import SentencePieceTrainer

os.makedirs("models", exist_ok=True)

SentencePieceTrainer.Train(
    "--input=datasets/corpus.txt "
    "--model_prefix=models/petition_bpe "
    "--vocab_size=8000 "
    "--model_type=bpe"
)

print("학습 완료!")
print("생성 파일: models/petition_bpe.model, models/petition_bpe.vocab")
```

```python
# 예제 5.13 BPE 토크나이저 사용 (SentencePieceProcessor)
from sentencepiece import SentencePieceProcessor

tokenizer = SentencePieceProcessor()
tokenizer.load("models/petition_bpe.model")

sentence = "안녕하세요, 토크나이저가 잘 학습되었군요!"
sentences = ["이렇게 입력값을 리스트로 받아서", "쉽게 토크나이저를 사용할 수 있답니다"]

# 하위 단어로 토큰화
tokenized_sentence  = tokenizer.encode_as_pieces(sentence)
tokenized_sentences = tokenizer.encode_as_pieces(sentences)

# 정수 인코딩
encoded_sentence  = tokenizer.encode_as_ids(sentence)
encoded_sentences = tokenizer.encode_as_ids(sentences)

# 디코딩 (정수 → 문장)
decode_ids    = tokenizer.decode_ids(encoded_sentences)
decode_pieces = tokenizer.decode_pieces(tokenized_sentences)

print("단일 문장 토큰화 :", tokenized_sentence)
print("여러 문장 토큰화 :", tokenized_sentences)
print("단일 문장 정수 인코딩 :", encoded_sentence)
print("정수 인코딩에서 문장 변환 :", decode_ids)
print("하위 단어에서 문장 변환 :", decode_pieces)
```

```python
# 예제 5.14 어휘 사전 불러오기
from sentencepiece import SentencePieceProcessor

tokenizer = SentencePieceProcessor()
tokenizer.load("models/petition_bpe.model")

vocab = {idx: tokenizer.id_to_piece(idx) for idx in range(tokenizer.get_piece_size())}

print("어휘 사전 샘플 (앞 5개):", list(vocab.items())[:5])
print("vocab size :", len(vocab))
print("\n※ <unk>=미등록 단어, <s>=문장 시작, </s>=문장 끝")
```

### 하위 단어 토큰화 — WordPiece (Hugging Face Tokenizers)
- 확률 기반으로 글자 쌍 병합 (BPE는 빈도 기반)
- 글자 쌍 병합 점수 = freq(xy) / (freq(x) × freq(y))
- 학습 결과를 JSON 파일로 저장

```python
!pip install tokenizers
```

```python
# 예제 5.15 WordPiece 토크나이저 학습
from tokenizers import Tokenizer
from tokenizers.models import WordPiece
from tokenizers.normalizers import Sequence, NFD, Lowercase
from tokenizers.pre_tokenizers import Whitespace
import os

os.makedirs("models", exist_ok=True)

tokenizer = Tokenizer(WordPiece())                        # 워드피스 모델 생성
tokenizer.normalizer = Sequence([NFD(), Lowercase()])     # 정규화: NFD + 소문자 변환
tokenizer.pre_tokenizer = Whitespace()                    # 사전 토큰화: 공백 기준 분리

tokenizer.train(["datasets/corpus.txt"])                  # 학습
tokenizer.save("models/petition_wordpiece.json")           # JSON으로 저장

print("WordPiece 토크나이저 학습 완료!")
print("저장 파일: models/petition_wordpiece.json")
```

```python
# 예제 5.16 WordPiece 토크나이저 사용
from tokenizers import Tokenizer
from tokenizers.decoders import WordPiece as WordPieceDecoder

tokenizer = Tokenizer.from_file("models/petition_wordpiece.json")
tokenizer.decoder = WordPieceDecoder()

sentence  = "안녕하세요, 토크나이저가 잘 학습되었군요!"
sentences = ["이렇게 입력값을 리스트로 받아서", "쉽게 토크나이저를 사용할 수 있답니다"]

encoded_sentence  = tokenizer.encode(sentence)
encoded_sentences = tokenizer.encode_batch(sentences)

print("인코딩 형식       :", type(encoded_sentence))
print("단일 문장 토큰화  :", encoded_sentence.tokens)
print("여러 문장 토큰화  :", [enc.tokens for enc in encoded_sentences])
print("단일 문장 정수 인코딩:", encoded_sentence.ids)
print("여러 문장 정수 인코딩:", [enc.ids for enc in encoded_sentences])
print("디코딩 결과       :", tokenizer.decode(encoded_sentence.ids))
```

## 6장. 임베딩 (~TF-IDF)

### 텍스트 벡터화 — 원-핫 인코딩 & 빈도 벡터화
- 원-핫 인코딩: 단어 색인 위치를 1로, 나머지를 0으로 표현
- 빈도 벡터화: 단어 등장 빈도수로 표현

```python
# 원-핫 인코딩 직접 구현
corpus = ["I like apples", "I like bananas"]

# 어휘 사전 구성
vocab = sorted(set(word for sent in corpus for word in sent.split()))
word2idx = {word: idx for idx, word in enumerate(vocab)}

print("어휘 사전:", word2idx)
print()

# 원-핫 인코딩
for sent in corpus:
    vec = [0] * len(vocab)
    for word in sent.split():
        vec[word2idx[word]] = 1
    print(f"{sent!r:25} → {vec}")

print("\n※ 단어 순서나 의미 정보가 반영되지 않음")
```

```python
# BoW (Bag of Words) — 빈도 벡터화
from collections import Counter

corpus = [
    "That movie is famous movie",
    "I like that actor",
    "I don't like that actor"
]

# 전체 어휘 사전
all_words = sorted(set(word.lower() for sent in corpus for word in sent.split()))
print("어휘 사전:", all_words)
print()

# 각 문서의 BoW 벡터
for sent in corpus:
    counter = Counter(word.lower() for word in sent.split())
    vec = [counter.get(word, 0) for word in all_words]
    print(f"{sent!r}")
    print(f"  BoW: {vec}\n")
```

### 예제 6.1 N-gram 구현
- N개의 연속된 단어 시퀀스를 하나의 단위로 취급
- N=1: 유니그램, N=2: 바이그램, N=3: 트라이그램
- 파이썬 직접 구현 / NLTK 라이브러리 모두 사용

```python
# 예제 6.1 N-gram 구현
import nltk

# 파이썬으로 직접 구현
def make_ngrams(sentence, n):
    words = sentence.split()
    ngrams = zip(*[words[i:] for i in range(n)])
    return list(ngrams)

sentence = "안녕하세요 만나서 진심으로 반가워요"

print("=== 파이썬 직접 구현 ===")
print("Unigram (N=1):", make_ngrams(sentence, 1))
print("Bigram  (N=2):", make_ngrams(sentence, 2))
print("Trigram (N=3):", make_ngrams(sentence, 3))

print("\n=== NLTK 라이브러리 사용 ===")
print("Unigram (N=1):", list(nltk.ngrams(sentence.split(), 1)))
print("Bigram  (N=2):", list(nltk.ngrams(sentence.split(), 2)))
print("Trigram (N=3):", list(nltk.ngrams(sentence.split(), 3)))
```

### 예제 6.2 TF-IDF 벡터화 (scikit-learn)
- TF(단어 빈도) × IDF(역문서 빈도) = 문서 내 단어의 중요도
- 자주 등장하지만 여러 문서에 걸쳐 드물게 나오는 단어가 높은 값
- 수식: IDF = log(전체 문서 수 / (DF + 1))

```python
!pip install scikit-learn
```

```python
# 예제 6.2 TF-IDF 벡터화
from sklearn.feature_extraction.text import TfidfVectorizer

corpus = [
    "That movie is famous movie",
    "I like that actor",
    "I don't like that actor"
]

tfidf_vectorizer = TfidfVectorizer()
tfidf_vectorizer.fit(corpus)                                # 단어 사전 구축 및 학습
tfidf_matrix = tfidf_vectorizer.transform(corpus)          # 벡터화

print("=== TF-IDF 행렬 (문서 수 × 단어 수) ===")
print(tfidf_matrix.toarray().round(4))

print("\n=== 단어 사전 (단어: 색인) ===")
print(tfidf_vectorizer.vocabulary_)

print("\n=== 각 단어의 IDF 값 ===")
for word, idx in sorted(tfidf_vectorizer.vocabulary_.items(), key=lambda x: x[1]):
    print(f"  {word:10}: IDF = {tfidf_vectorizer.idf_[idx]:.4f}")
```

```python
# TF-IDF 상위 단어 추출 — 각 문서에서 중요한 단어 파악
import numpy as np
from sklearn.feature_extraction.text import TfidfVectorizer

corpus = [
    "That movie is famous movie",
    "I like that actor",
    "I don't like that actor"
]

tfidf_vectorizer = TfidfVectorizer()
tfidf_matrix = tfidf_vectorizer.fit_transform(corpus)  # fit + transform 한번에

# 색인 → 단어 역매핑
idx2word = {v: k for k, v in tfidf_vectorizer.vocabulary_.items()}

print("=== 각 문서에서 TF-IDF 상위 3개 단어 ===")
for doc_idx, doc in enumerate(corpus):
    scores = tfidf_matrix[doc_idx].toarray()[0]
    top3_idx = np.argsort(scores)[::-1][:3]
    top3_words = [(idx2word[i], round(scores[i], 4)) for i in top3_idx]
    print(f"\n문서 {doc_idx+1}: '{doc}'")
    print(f"  중요 단어: {top3_words}")

print("\n※ TF-IDF 한계: 단어 순서/문맥 미반영 → Word2Vec 등 밀집 표현으로 보완")
```

### TfidfVectorizer 주요 매개변수 실험

```python
# N-gram 범위 설정 예시
from sklearn.feature_extraction.text import TfidfVectorizer

corpus = [
    "That movie is famous movie",
    "I like that actor",
    "I don't like that actor"
]

# 유니그램만
tfidf_uni = TfidfVectorizer(ngram_range=(1, 1))
tfidf_uni.fit(corpus)

# 유니그램 + 바이그램
tfidf_bi = TfidfVectorizer(ngram_range=(1, 2))
tfidf_bi.fit(corpus)

print("유니그램 어휘 사전:", list(tfidf_uni.vocabulary_.keys()))
print(f"  → 어휘 수: {len(tfidf_uni.vocabulary_)}개")
print()
print("유니그램+바이그램 어휘 사전:")
print(list(tfidf_bi.vocabulary_.keys()))
print(f"  → 어휘 수: {len(tfidf_bi.vocabulary_)}개")
```

## 정리

| 방법 | 라이브러리 | 특징 |
|------|-----------|------|
| 단어 토큰화 | 내장 `split()` | 가장 간단, OOV 문제 |
| 글자 토큰화 | 내장 `list()` | 작은 어휘 사전, 긴 시퀀스 |
| 자소 토큰화 | `jamo` | 한글 자음·모음 단위 분리 |
| 형태소(한국어) | `KoNLPy` | Okt/Kkma, 문법 구조 반영 |
| 형태소(영어) | `NLTK`, `spaCy` | 품사 태깅, 구문 분석 |
| BPE | `SentencePiece` | 빈도 기반 하위 단어, OOV 해결 |
