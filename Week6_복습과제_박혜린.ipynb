# **U-Net 모델 구현**

✅ 아래 코드는 PyTorch로 작성한 간단한 U-Net 모델입니다.

✅ 주요 목표: 작은 이미지 세트로 segmentation 학습을 테스트하기!

✅ 특징: 업샘플링(deconv/upsample 선택 가능), 입력 스케일링, Gaussian noise 추가, 배치 정규화 적용

```python

import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import DataLoader, TensorDataset

# --- 작은 U-Net 정의 ---
class UNetSmall(nn.Module):
    def __init__(self, upsample_mode='deconv', net_scaling=None, edge_crop=0, gaussian_noise_std=0.1):
        super().__init__()
        self.net_scaling = net_scaling
        self.edge_crop = edge_crop
        self.gaussian_noise_std = gaussian_noise_std
        self.upsample_mode = upsample_mode

         # Convolution Block: Conv -> ReLU -> Conv -> ReLU
        def conv_block(in_c, out_c):
            return nn.Sequential(
                nn.Conv2d(in_c, out_c, 3, padding=1),
                nn.ReLU(inplace=True),
                nn.Conv2d(out_c, out_c, 3, padding=1),
                nn.ReLU(inplace=True)
            )

        # 인코더 부분 목표 -> 128 까지!
        self.enc1 = conv_block(3, 16)
        self.enc2 = conv_block(16, 32)
        self.enc3 = conv_block(32, 64)
        self.enc4 = conv_block(64, 128)
        # Bottleneck
        self.center = conv_block(128, 128)
        self.pool = nn.MaxPool2d(1)


        # 업샘플링 부분 정의 (deconv 또는 upsample+conv) 로
        if upsample_mode == 'deconv':
            def upsample_layer(in_c, out_c):
                return nn.ConvTranspose2d(in_c, out_c, 1)
        else:
            def upsample_layer(in_c, out_c):
                return nn.Sequential(
                    nn.Upsample(scale_factor=2, mode='nearest'),
                    nn.Conv2d(in_c, out_c, 1)
                )
        # 디코더 부분
        self.up4 = upsample_layer(128, 128)
        self.dec4 = conv_block(128+128, 128)
        self.up3 = upsample_layer(128, 64)
        self.dec3 = conv_block(64+64, 64)
        self.up2 = upsample_layer(64, 32)
        self.dec2 = conv_block(32+32, 32)
        self.up1 = upsample_layer(32, 16)
        self.dec1 = conv_block(16+16, 16)

        # 최종 출력: 1채널(segmentation mask)
        self.final = nn.Conv2d(16,1,1)

    def forward(self, x):
        if self.net_scaling:
        # 입력 스케일링
            x = F.avg_pool2d(x, self.net_scaling)
        # Gaussian noise 추가
        if self.gaussian_noise_std:
            noise = torch.randn_like(x) * self.gaussian_noise_std
            x = x + noise

        # 배치 정규화 적용
        x = F.batch_norm(x, running_mean=None, running_var=None, training=True)

        # 인코더
        e1 = self.enc1(x)
        e2 = self.enc2(self.pool(e1))
        e3 = self.enc3(self.pool(e2))
        e4 = self.enc4(self.pool(e3))
        # Bottleneck
        center = self.center(e4)

        # 디코더 + Skip Connection
        d4 = self.up4(center)
        d4 = torch.cat([d4,e4], dim=1)
        d4 = self.dec4(d4)

        d3 = self.up3(d4)
        d3 = torch.cat([d3,e3], dim=1)
        d3 = self.dec3(d3)

        d2 = self.up2(d3)
        d2 = torch.cat([d2,e2], dim=1)
        d2 = self.dec2(d2)

        d1 = self.up1(d2)
        d1 = torch.cat([d1,e1], dim=1)
        d1 = self.dec1(d1)

        out = self.final(d1)
        out = torch.sigmoid(out)
         # Crop + Pad
        if self.edge_crop > 0:
            out = out[:, :, self.edge_crop:-self.edge_crop, self.edge_crop:-self.edge_crop]
            out = F.pad(out, (self.edge_crop, self.edge_crop, self.edge_crop, self.edge_crop))
        # Upsample 다시 확대
        if self.net_scaling:
            out = F.interpolate(out, scale_factor=self.net_scaling, mode='nearest')

        return out
```

## **학습용 데이터 생성하기**

```python
batch_size = 32
x = torch.randn(batch_size, 3, 128, 128)  # RGB 이미지
y = (torch.rand(batch_size, 1, 128, 128) > 0.5).float()  # 랜덤 이진 마스크

dataset = TensorDataset(x, y)
loader = DataLoader(dataset, batch_size=2, shuffle=True)
```

## **데이터 시각화**

```python
import matplotlib.pyplot as plt
# 첫 번째 이미지를 시각화
sample_x = x[0]
sample_y = y[0]

plt.figure(figsize=(8, 4))
plt.subplot(1, 2, 1)
plt.imshow(sample_x.permute(1, 2, 0).numpy())
plt.title('Input Image')

plt.subplot(1, 2, 2)
plt.imshow(sample_y.squeeze().numpy(), cmap='gray')
plt.title('Ground Truth Mask')
plt.show()
```

## **모델 선언 및 학습**

```python
# --- 모델 선언 및 학습 ---
model = UNetSmall(upsample_mode='deconv', net_scaling=None, edge_crop=0)
optimizer = torch.optim.Adam(model.parameters(), lr=1e-3)
criterion = nn.BCELoss()

# 간단히 30 epoch만
for epoch in range(10):
    model.train()
    total_loss = 0
    for xb, yb in loader:
        pred = model(xb)
        loss = criterion(pred, yb)
        optimizer.zero_grad()
        loss.backward()
        optimizer.step()
        total_loss += loss.item()

    print(f"Epoch {epoch+1}, Loss: {total_loss/len(loader):.4f}")
```

[답변]


---

[DeepLab v1]
- 일반적인 CNN은 풀링(Pooling)을 거치며 이미지 해상도가 줄어드는데, 이를 방지하기 위해 필터 사이에 빈 공간을 두는 'Atrous Convolution'을 처음 도입했습니다.

- 한계: 경계선이 뭉개지는 문제를 해결하기 위해 후처리 기법인 CRF를 사용했는데, 계산량이 많고 복잡하다는 단점이 있었습니다.

[DeepLab v2]
- 다양한 크기의 객체를 잘 잡아내기 위해, 서로 다른 확장 비율(dilation rate)을 가진 Atrous Convolution 여러 개를 병렬로 연결했습니다. 이를 통해 다각도의 문맥(Context) 정보를 동시에 파악할 수 있게 되었습니다.

- 백본: VGG-16 외에도 ResNet 등을 백본 네트워크로 사용하기 시작했습니다.

[DeepLab v3]
- 기존 ASPP에 1x1 Convolution과 Global Average Pooling을 추가하여 전역적인 정보를 더 잘 반영하게 했습니다.

= 후처리 과정이었던 CRF 없이도 높은 성능을 낼 수 있도록 아키텍처를 최적화했습니다.

- Atrous Convolution을 직렬로 연결하는 'Multi-grid' 방식을 도입했습니다.

[DeepLab v3+]
- Encoder-Decoder: DeepLabv3를 Encoder로 사용하고, 저수준 특징(Low-level features)을 결합하는 간단한 Decoder를 추가해 물체의 경계선(Boundary)을 훨씬 정교하게 복원했습니다.

- Atrous Separable Convolution: Depthwise Separable Convolution에 Atrous를 결합하여 연산량을 획기적으로 줄이면서도 성능은 유지하거나 높였습니다.

- 현재 가장 대중적으로 많이 쓰이는 완성도 높은 버전입니다.
